```python
# Calculate and draw the path of a robot given a heading.  Draw also the
# shortest paths to the nearest horizontal and vertical walls.

import random
import math
import matplotlib.pyplot as plt 
  # Use plt as an alias so we don't have to type out matplolib.pyplot every time

# Set up the window
plt.close('all')        # close all currently open figure windows
plt.figure()            # creates a new figure window
ax= plt.gca()           # gets current axes
ax.set_aspect('equal')  # unit lengths on x- and y-axis are equal (equal scaling)

# Parameter
side= 10.    # side length of the square room

# Draw walls in blue, lower-left corner at (0,0)
plt.axis((-1, side+1, -1, side+1))  # x-axis limits -1 to side+1, 
                                    # y-axis limits -1 to side+1
plt.plot([0, side],    [0, 0], 'b-', linewidth=3)  # format: Blue line of width 3
plt.plot([side, side], [0, side], 'b-', linewidth=3)
plt.plot([side, 0],    [side, side], 'b-', linewidth=3)
plt.plot([0, 0],       [side, 0], 'b-', linewidth=3)

### Do not change the code above ###
####################################
### Modify the code below ##########

# Location of Robot
#xr= 10*random.random()
xr=6
yr= 10*random.random()

plt.plot(xr, yr, 'ro')  # format: Red circle
plt.text(xr, yr, '  Robot')

# User-clicked point
plt.title("Click away from the robot to set the robot's heading")
plt.show()  # show the plot
point= plt.ginput(1)  # Accepts one user click; note time-out of 30 seconds
#xu, yu= point[0]      # Stores x-coord of click in variable xu,
xu=6#        y-coord of click in variable yu
plt.plot(xu, yu, 'm+')   # format: Magenta cross
plt.title(f'You clicked at ({xu:.1f},{yu:.1f})')

# Draw line to connect the two points
plt.plot([xr, xu], [yr, yu], 'k:')  # format: blacK dotted line of default width

#Caculate the heading#
plt.show()
heading = math.atan2(yu-yr, xu-xr)

###straight-line path to a wall 

#Use two if statements to separate the top part and bottom part. If the user 
#clicks above the robot point, Python will use the first if statement.If the 
#user clicks below the robot point, Python will use the second if statement.
#If it alligns with the robot in y=axis it will use the third if statement.
#I use similar triangle to determine the endpoints. Written as [a,b]. Python 
#needs to figure out a and b by calculating the portportion T.This calculation 
#separates the user click point into three part, if the user's click is near the 
#left(which means its radian is larger than the radians form by left top 
#endpoint and the robot point)then the line will touch the left wall and use
#the T caculation for left wall. Same thing for up and right.

#naming
dy = yu - yr
dx = xu - xr

#if the user clicks above the robot
if dy>0 :     
    if math.atan2(10-yr, -xr)<heading: #left
        T=(-xr)/(dx)
    elif math.atan2(10-yr,10-xr)>heading: #right
        T=(10-xr)/(dx)
    else: #up
        T=(10-yr)/(dy)

#if the user clicks below the robot

#Same reason as the last one.
if dy<0:
    if math.atan2(-yr,-xr)>heading:#left
        T=(-xr)/(dx)
    elif math.atan2(-yr,10-xr)<heading:#right
        T=(10-xr)/(dx)
    else: #down
        T=(-yr)/(dy) 

#The T is the portportion of change in length and width
a=xr+(T)*(dx)
b=yr+(T)*(dy)

#if the it alligns with y-axis, it will use this statement
#override the above statement to connect the wall horizontally.
if dy==0 and xu>xr:
    a=10
    b=yr
if dy==0 and xu<xr:
    a=0
    b=yr

#convert heading to radian and print it
heading = heading%(2*math.pi)

plt.title(f"The robot's heading is {heading:0.1f} radians" )

#draw[a,b] and connect robot point to [a,b]
plt.plot(a,b,'ko')
plt.plot([xr, a], [yr, b], 'r-') 

#shortest distance to horizontal wall
if 10-yr < yr:
    plt.plot([xr,xr], [yr,side],"p:")
else:
    plt.plot([xr,xr],[yr,0], "p:")
#shortest distance to vertical wall
if 10-xr < xr:
    plt.plot([xr,side],[yr,yr],"c--")
else:
    plt.plot([xr,0],[yr,yr],"c--")

### Do not change the code below ###
print('Bye')
```
